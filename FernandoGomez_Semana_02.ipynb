*   **Pregunta 1.**

Busca y elimina todos los saltos de línea '\n', así como las etiquetas de 0s y 1s que se encuentran al final de cada comentario. En esta actividad solo trabajaremos con la limpieza de datos, por lo que no requeriremos dichas etiquetas por el momento.

Una vez finalizado, imprime los primeros 10 comentarios del resultado obtenido.


```python
# Lista para almacenar los comentarios limpios
docs_limpios = []

for comentario in docs:
    # Utilizar expresiones regulares para eliminar el salto de línea '\n'
    # y la etiqueta de fin de línea (ej. espacio seguido de 0 o 1).
    limpio = re.sub(r'\s+[01]\n', '', comentario)
    # En caso de que el salto de línea esté solo o queden espacios sueltos al final
    limpio = limpio.strip()
    docs_limpios.append(limpio)

# Verificar que la lista contenga 1000 elementos
print("Longitud de la lista limpia:", len(docs_limpios))

```

*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!".

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen.

Indica cuántos resultados obtuviste.



```python
# Lista para acumular los resultados
resultados = []

# Expresión regular para encontrar palabras seguidas de dos o más signos de admiración
# \b\w+ captura la palabra completa y \!{2,} captura dos o más signos de admiración
patron = re.compile(r'\b\w+\!{2,}')

# Iterar sobre la lista de comentarios limpios
for comentario in docs_limpios:
    # Buscar todas las coincidencias en el comentario
    encontrados = patron.findall(comentario)
    if encontrados:
        resultados.extend(encontrados)
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras (caracteres alfabéticos) que inician con una letra mayúscula y terminan con una minúscula, pero donde además dicha palabra no sea la primera palabra del comentario (string).

Indica la cantidad de resultados obtenidos.

```python
# Lista para almacenar las palabras encontradas
palabras_cumplen = []

# Expresión regular para encontrar palabras que inicien con mayúscula y terminen con minúscula
# [A-Z] para iniciar con mayúscula y [a-z]+ para terminar con letras minúsculas (longitud mínima 2 caracteres)
patron = re.compile(r'\b[A-Z][a-z]+\b')

# Iterar sobre cada comentario limpio
for comentario in docs_limpios:
    # Dividir el comentario en palabras respetando los espacios
    palabras_en_comentario = comentario.split()
    
    # Asegurar de que haya al menos dos palabras para evaluar
    if len(palabras_en_comentario) > 1:
        # Evaluar desde la segunda palabra en adelante (índice 1 en adelante)
        for palabra in palabras_en_comentario[1:]:
            if patron.fullmatch(palabra):
                # Limpiar posibles signos de puntuación pegados a la palabra
                palabra_limpia = re.sub(r'[^a-zA-Z]', '', palabra)
                if len(palabra_limpia) > 1 and palabra_limpia[0].isupper() and palabra_limpia[-1].islower():
                    palabras_cumplen.append(palabra_limpia)

# Cantidad de resultados
print(f"Total de resultados obtenidos: {len(palabras_cumplen)}\n")
```

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos, incluyendo algún tipo de acentuación. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
# Lista para almacenar los comentarios limpios finales
docs_procesados = []

for comentario in docs_limpios:
    # Paso 1: Eliminar signos de puntuación y caracteres especiales, conservando letras y acentos.
    # [^\w\s] elimina puntuación, y podemos refinar para mantener letras y vocales acentuadas.
    # Una forma segura con 're' es reemplazar todo lo que no sea letra o espacio por espacio o nada:
    # En este caso, reemplazar caracteres que no sean letra (incluyendo acentuadas) o espacio.
    comentario_limpio = re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚ\s]', '', comentario)
    
    # Paso 2: Transformar todos los caracteres alfabéticos a minúsculas
    comentario_limpio = comentario_limpio.lower()
    
    # Paso 3: Eliminar espacios en blanco adicionales y espacios al inicio/final
    # 'split()' seguido de 'join' elimina cualquier espacio múltiple y los extremos
    comentario_limpio = " ".join(comentario_limpio.split())
    
    docs_procesados.append(comentario_limpio)
```

```python
# Primeros 10 comentarios resultantes
print("Primeros 10 comentarios limpios y procesados:\n")
for idx, comentario in enumerate(docs_procesados[:10], 1):
    print(f"{idx}. {comentario}")
```
